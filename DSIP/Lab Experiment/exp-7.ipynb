```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
def generate_histogram(image):
    histogram = cv2.calcHist(
        [image],
        [0],
        None,
        [256],
        [0,256]
    )
    return histogram
```

```python
input_path =(r"C:\Users\admin\Documents\DSIP\Lab Experiment\low constract image.jpeg")
input_image = cv2.imread(input_path)

gray_image = cv2.cvtColor(input_image,cv2.COLOR_BGR2GRAY)

input_histogram = generate_histogram(gray_image)

intensity_values = np.arange(256)
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(gray_image, cmap="gray")
plt.title("Gray Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.bar(intensity_values, input_histogram.ravel(), color="black")
plt.title("Input Image Histogram")
plt.xlabel("INTENSITY")
plt.ylabel("Frequency / number of pixels")
plt.xlim([0, 255])
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

```python
total_pixels = gray_image.size

normalized_histogram = input_histogram / total_pixels

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.bar(intensity_values, input_histogram.ravel(), color="black")
plt.title('Normalized Histogram')
plt.xlabel('Intensity Value')
plt.ylabel('Normalized Frequency')
plt.xlim([0, 255])
```
