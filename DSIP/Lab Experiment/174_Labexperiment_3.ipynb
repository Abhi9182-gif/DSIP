```python
import numpy as np

x = np.array([1, 2, 3, 4, 5])
h = np.array([2, 4, 6, 8, 10])
```

```python
# Cross-correlation
r_xh = np.correlate(x, h, mode='full')

print("Cross-Correlation r_xh(m):")
print(r_xh)
```

```python
# Auto-correlation of x
r_xx = np.correlate(x, x, mode='full')

print("\nAuto-Correlation r_xx(m):")
print(r_xx)
```

```python
lags = np.arange(-len(x)+1, len(x))

print("\nLags:")
print(lags)

print("\nCross-Correlation with lags:")
for lag, val in zip(lags, r_xh):
    print(f"m = {lag} : {val}")
```
