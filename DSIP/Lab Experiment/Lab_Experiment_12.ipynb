**Aim:** To simulate the Hit-or-Miss transformation on binary images using Python and OpenCV for detecting specific patterns and shapes.

## Import Required Libraries

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

## Load and Display the Original Image

```python
image = cv2.imread(
    r"C:\Users\admin\Documents\DSIP\Lab Experiment\dark_car.png",
    cv2.IMREAD_GRAYSCALE
)

if image is None:
    raise FileNotFoundError("Image not found. Check the image path.")

plt.figure(figsize=(8, 6))
plt.imshow(image, cmap="gray")
plt.title("Original Grayscale Image")
plt.axis("off")
plt.show()
```

## Convert the Image into Binary Format

```python
_, binary = cv2.threshold(
    image,
    127,
    1,
    cv2.THRESH_BINARY
)

plt.figure(figsize=(8, 6))
plt.imshow(binary, cmap="gray")
plt.title("Binary Image")
plt.axis("off")
plt.show()
```

## Define the Structuring Element
- `1` = foreground pixels
- `-1` = background pixels
- `0` = don't-care pixels

```python
pattern = np.array([
    [-1,  1, -1],
    [ 1,  1,  1],
    [-1,  1, -1]
], dtype=np.int8)

print("Structuring Element:")
print(pattern)
```

## Apply Hit-or-Miss Transformation

```python
result = cv2.morphologyEx(
    binary,
    cv2.MORPH_HITMISS,
    pattern
)

print("Hit-or-Miss Transformation Completed!")
```

## Display the Hit-or-Miss Output

```python
plt.figure(figsize=(8, 6))

plt.imshow(result * 255, cmap="gray")

plt.title("Hit-or-Miss Transformation")
plt.axis("off")

plt.show()
```

## Compare Original, Binary and Transformed Images

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(binary, cmap="gray")
plt.title("Binary Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(result * 255, cmap="gray")
plt.title("Hit-or-Miss Output")
plt.axis("off")

plt.tight_layout()
plt.show()
```
